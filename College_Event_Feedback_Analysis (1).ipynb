## 3. Rating Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('college_event_feedback (1).csv')
df.groupby('Event_Name')['Rating'].mean().plot(kind='bar')
plt.title('Average Rating per Event')
plt.show()
```

## 4. Sentiment Analysis

```python
from textblob import TextBlob
df['Sentiment_Score'] = df['Feedback_Text'].apply(lambda x: TextBlob(x).sentiment.polarity)
df['Sentiment'] = df['Sentiment_Score'].apply(lambda x: 'Positive' if x > 0 else 'Negative' if x < 0 else 'Neutral')
df.head()
```

## 5. Sentiment Distribution

```python
sns.countplot(x='Sentiment', data=df)
plt.title('Sentiment Distribution')
plt.show()
```

## 6. Rating Distribution

```python
sns.histplot(df['Rating'], bins=5)
plt.title('Rating Distribution')
plt.show()
```
